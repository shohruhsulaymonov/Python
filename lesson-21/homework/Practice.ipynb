```python
data3 = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank', 'Grace', 'Hank', 'Ivy', 'Jack'],
    'Department': ['HR', 'IT', 'Marketing', 'IT', 'Finance', 'HR', 'Marketing', 'IT', 'Finance', 'Marketing'],
    'Salary': [60000, 75000, 65000, 80000, 70000, 72000, 68000, 78000, 69000, 76000],
    'Experience (Years)': [3, 5, 2, 8, 4, 6, 3, 7, 2, 5]
}

employees = pd.DataFrame(data3)
```

```python
#9. Calculate the average salary for each department.
dep_avg_salary = employees.groupby('Department')['Salary'].mean().round(2)
```

```python
#10. Find the employee with the most experience.
max_exp = employees['Experience (Years)'].max()
employees[employees['Experience (Years)'] == max_exp]
```

```python
#11. Create a new column 'Salary Increase' representing the percentage increase in salary from the minimum salary in the dataframe.
min_salary = employees['Salary'].min()
employees['Salary increase'] = (100*(employees['Salary'] - min_salary)/min_salary).round(2)
```

```python
#12 Plot a bar chart to visualize the distribution of employees across different departments.
department_distribution = employees.groupby('Department')['Employee_ID'].nunique()
department_distribution.plot(kind='bar', color='red')
plt.title('Distribution of Employees')
plt.show()
```

```python
data4 = {
    'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Customer_ID': [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    'Product': ['A', 'B', 'A', 'C', 'B', 'C', 'A', 'C', 'B', 'A'],
    'Quantity': [2, 3, 1, 4, 2, 3, 2, 5, 1, 3],
    'Total_Price': [120, 180, 60, 240, 160, 270, 140, 300, 90, 180]
}

orders = pd.DataFrame(data4)
```

```python
#13. Calculate the total revenue from all orders.


total_revenue = orders['Total_Price'].sum()
```

```python
#14. Find the most ordered product.
most_ordered_product = orders['Product'].mode()
```

```python
#15. Calculate the average quantity of products ordered.
avg_qty = orders['Quantity'].mean()
```

```python
#16. Plot a pie chart to visualize the distribution of sales across different products.
products_grp = orders.groupby('Product')['Total_Price'].sum()

idx = products_grp.index
vals = products_grp.values

plt.pie(vals, labels=idx,  colors=['yellow', 'red', 'green'], autopct='%1.1f%%', wedgeprops={'edgecolor': 'black'})

plt.tight_layout()

plt.title('Product Income Dustribution')

plt.show()
```
